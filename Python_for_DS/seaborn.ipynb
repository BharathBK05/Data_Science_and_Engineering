```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#load the pre build datasets in seaborn
df = sns.load_dataset('tips')
df.head()
```

```python
#Basic ploting
#Scatter plot - relationship between 2 numerical values
sns.scatterplot(x='total_bill',y='tip',data=df)
plt.title('Total bill vs Tip')
```

```python
#Line plot --> Trends over time or sequence
sns.lineplot(x='size',y='tip',data=df)
plt.title('Line plot of size vs Tip')
```

```python
#Bar plot -->compare categorical groups with confidence intervals
sns.barplot(x='day',y='total_bill',data=df)
plt.title('Average bill by day')
```

```python
#Count plot - Frequency of category
sns.countplot(x='day',data=df)
plt.title('count of records by day')
```

```python
#Box plot --> show the distribution,outliers, quartilies
sns.boxplot(x='day',y='total_bill',data=df)
plt.title('Box plot of Bills by day')
```

```python
#violin plot --> shows density + box plot
sns.violinplot(x='day',y='total_bill',data=df)
```

```python
#strip plot
sns.stripplot(x='day',y='total_bill',data=df,jitter=True)
```

```python
#swarmplot
sns.swarmplot(x='day',y='total_bill',data=df)
```

```python
#Relation plot
sns.relplot(x='total_bill',y="tip",hue='sex',col='time',data=df)
```

```python
#Regression plot
sns.regplot(x='total_bill',y='tip',data=df)
```

```python
#lmplot (advanced regression)
sns.lmplot(x='total_bill',y='tip',hue='sex',col='time',data=df)
```
