```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
```

```python
#Block Diagram
#Input --> Process --> output

fig,ax = plt.subplots()

#draw rectangle
ax.add_patch(patches.Rectangle((0.1,0.4),0.2,0.2, edgecolor='black',facecolor='lightblue'))
ax.add_patch(patches.Rectangle((0.4,0.4),0.2,0.2, edgecolor='black',facecolor='lightgreen'))
ax.add_patch(patches.Rectangle((0.7,0.4),0.2,0.2, edgecolor='black',facecolor='lightcoral'))

#add labels
ax.text(0.2,0.5,'Input',ha='center',va='center',fontsize=12)
ax.text(0.5,0.5,'Process',ha='center',va='center',fontsize=12)
ax.text(0.8,0.5,'Ouput',ha='center',va='center',fontsize=12)

#add arrows
#ax.arrow(x,y,dx,dy)
ax.arrow(0.3,0.5,0.1,0, head_width= 0.05, head_length=0.02)
ax.arrow(0.6,0.5,0.1,0, head_width= 0.05, head_length=0.02)

#final touches
ax.set_xlim(0,1)
ax.set_ylim(0,1)
ax.axis('off')
```

```python
#drawing a carnot engine

fig,ax = plt.subplots()

#add heat source and sink
ax.text(0.5,0.9,"Hot Reservoir (T1)",ha='center',fontsize=12)
ax.text(0.5,0.1,"Cold Reservoir (T2)",ha='center',fontsize=12)

#add engine box
engine = patches.Rectangle((0.4,0.4),0.2,0.2,edgecolor='black',facecolor='orange')
ax.add_patch(engine)
ax.text(0.5,0.5,"Engine",ha='center',va='center',fontsize=12)

#add arrows
#heat input
ax.arrow(0.5,0.85,0,-0.15,head_width=0.03,head_length=0.03)
ax.text(0.52,0.75,'Q1',color="red")

#work output
ax.arrow(0.6,0.5,0.15,0,head_width=0.03,head_length=0.03)
ax.text(0.68,0.52,'W',color="blue")

#heat rejected
ax.arrow(0.5,0.4,0,-0.15,head_width=0.03,head_length=0.03)
ax.text(0.52,0.3,'Q2',color="green")

#Final touches
ax.set_xlim(0,1)
ax.set_ylim(0,1)
ax.axis('on')
plt.grid()
plt.title('Carnot Engine Schematic Diagram',fontsize=14)
```

```python
from re import A
#Custom shapes

from matplotlib.patches import Circle, FancyArrow, Arc

fig,ax = plt.subplots()

#add circle
ax.add_patch(Circle((0.3,0.5),0.05,color='skyblue'))
ax.text(0.3,0.5,'A',ha='center',va='center')

ax.add_patch(Circle((0.7,0.5),0.05,color='lightgreen'))
ax.text(0.7,0.5,'B',ha='center',va='center')

#connect with arrow
ax.add_patch(FancyArrow(0.35,0.5,0.25,0, width=0.01,head_width=0.03,head_length=0.02))

#add arc
ax.add_patch(Arc((0.5,0.5),0.4,0.3, theta1=0,theta2=180,linestyle='--'))

#Final touches
ax.set_xlim(0,1)
ax.set_ylim(0,1)
ax.axis('off')
plt.title('Custom Schematic Diagram',fontsize=14)
```

```python
#Circuit Diagram
fig,ax = plt.subplots()

#draw wires
ax.plot([0.1,0.9],[0.8,0.8],color='black')
ax.plot([0.1,0.1],[0.8,0.2],color='black')
ax.plot([0.1,0.9],[0.2,0.2],color='black')
ax.plot([0.9,0.9],[0.8,0.2],color='black')

#add resistor
resistor = patches.Rectangle((0.4,0.75),0.2,0.2,edgecolor='black',facecolor='gray')
ax.add_patch(resistor)
ax.text(0.5,0.8,'R',ha='center',va='center',fontsize=12)

#Add battery symbol
ax.plot([0.85,0.85],[0.35,0.45],color='black',linewidth=2) #long
ax.plot([0.87,0.87],[0.38,0.42],color='black',linewidth=2) #short
ax.text(0.86,0.3,'battery',ha='center')

#Final touches
ax.set_xlim(0,1)
ax.set_ylim(0,1)
ax.axis('off')
plt.title('Circuit Diagram',fontsize=14)
```
