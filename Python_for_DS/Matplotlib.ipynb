```python
import matplotlib.pyplot as plt

#basic plotting
x = [1,2,3,4]
y = [10,20,25,30]

plt.plot(x,y)
plt.title('Basic Line plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
#Line plot
plt.plot([1,2,3],[2,4,1], color='green', linestyle='--',marker='o')
plt.title('Line plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
```

```python
#Bar chart
categories = ['Male','Female','Others']
values = [10,25,17]

plt.bar(categories, values, color='orange')
plt.title('Bar Chart')
plt.xlabel('Category')
plt.ylabel('Counts')
```

```python
#Histogram
import numpy as np

data = np.random.randn(1000)

plt.hist(data, bins=30)
plt.title('Histogram')
```

```python
#Scatter plot - distribution
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x,y)
plt.title('Scatter Plot')
```

```python
#Pie chart
size = [25,35,20,20]
labels = ['Python','Java','C++','C']

plt.pie(size,labels=labels,autopct='%1.1f%%')
plt.title('Pie Chart')
```

```python
#Multiple plots in one figure
plt.subplot(1,2,1) #1 row , 2 columns , 1st plot
plt.plot([1,2,3],[1,4,9])
plt.title('1st plot')

plt.subplot(1,2,2)
plt.plot([1,2,3],[9,4,1])
plt.title('2st plot')

plt.tight_layout()
```

```python
#multiple plots in same graph
plt.plot([1,2,3],[2,4,1], color='green', linestyle='--',marker='o')
plt.plot([2,3,1],[1,2,3],color='red')
plt.title('Line plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
```

```python
#line plot from dataframe

import pandas as pd

df = pd.DataFrame(
    {
        'Year':[2018,2019,2020,2021],
        'Sales':[100,150,200,250]
    }
)

plt.plot(df['Year'],df['Sales'])
plt.title('Line plot')
plt.xlabel('Year')
plt.ylabel('Sales')
```

```python
#Advanced Styling
plt.figure(figsize=(8,3))
plt.plot([1,2,3],[1,4,9])
```

```python
#save figures
plt.plot([1,2,3],[1,4,9])
plt.savefig('plot.png')
```

```python
#Time series plot
date_range = pd.date_range(start='1/1/2020', end='1/10/2020',freq='D')
df = pd.DataFrame(date_range,columns=['date'])

df['sales'] = np.random.randint(0,100,size=len(date_range))

df.set_index('date',inplace=True)

df.plot()
plt.title('Random Time series data')
```
