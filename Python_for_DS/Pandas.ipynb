Level 3 - Advanced Operations

```python
#Groupby and aggregation
df.groupby('RainToday')[['Humidity9am','Humidity3pm']].mean()
df.groupby('RainToday')[['Humidity9am','Humidity3pm']].sum()
df.groupby('RainToday')[['Humidity9am','Humidity3pm']].count()
```

```python
#Merge
city = pd.DataFrame({
    'Location':['Sydney','Melbourne','Delhi'],
    'Region':['NSW','VIC','ABC']
})



country = pd.DataFrame({
    'Location':['Sydney','Melbourne','Mumbai'],
    'Country':['Australia','New Zealand','India']
})

df = pd.merge(city,country,on='Location',how='inner')
df

#inner join --> it will takes only matching values from both tables
#left join ---> it will take all the values from the left table and only matched values from the right
#right join ---> it takes all the vlaues from the right and only mathing values from the left
```

```python
#how to concat the dataframes
#concat --> rows

df1 = df.iloc[0:5]
df2 = df.iloc[6:10]

#combine
df_combined = pd.concat([df1,df2],axis=0)
df_combined.head(10)

```

```python
#concat ---> columns
temp = df[['MinTemp','MaxTemp']]
humidity = df[['Humidity9am','Humidity3pm']]

df_col_combined = pd.concat([temp,humidity],axis=1)
df_col_combined.head()
```

```python
#Lambda
df['TempCategory'] = df['MaxTemp'].apply(lambda x: 'Hot' if x > 30 else 'Mild')
df['TempCategory'].head(20)
```

```python
def find_category(temp):
  if temp > 30:
    return 'Hot'
  else:
    return 'Mild'
```

```python
df['TempCategory_1'] = df['MaxTemp'].apply(find_category)
df['TempCategory'].head(2)
```
