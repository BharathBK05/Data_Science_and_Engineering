```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Load the data
df = pd.read_csv('/content/owid-covid-data.csv')
df.head()
```

```python
#Filtering
india = df.loc[df['location']=='India']
india.head()
```

```python
india.info()
```

```python
#convert to date
india['date'] = pd.to_datetime(india['date'])
```

```python
india.info()
```

```python
#Plot new cases over time
plt.figure(figsize=(12,6))
sns.lineplot(data=india,x='date',y='new_cases')
```

```python
#Total vaccination Trend
sns.lineplot(data=india,x='date',y='total_vaccinations')
```

```python
#Moving average
india['7_day_avg'] = india['new_cases'].rolling(window=7).mean()
sns.lineplot(data=india,x='date',y='7_day_avg')
```
